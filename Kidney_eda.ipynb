```python
import os
from PIL import Image
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
DATASET_PATH = "data/Kidney Cancer"

classes = [
    "Normal",
    "Tumor"
]

print("Dataset path:", DATASET_PATH)
print("Classes:", classes)
```

```python
class_counts = {}

for cls in classes:
    class_path = os.path.join(DATASET_PATH, cls)

    files = [
        file for file in os.listdir(class_path)
        if os.path.isfile(os.path.join(class_path, file))
    ]

    class_counts[cls] = len(files)

print("Class distribution:")

for cls, count in class_counts.items():
    print(f"{cls}: {count}")

print("\nTotal images:", sum(class_counts.values()))
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    class_counts.keys(),
    class_counts.values()
)

plt.title("Kidney Cancer Dataset - Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Images")

for i, count in enumerate(class_counts.values()):
    plt.text(
        i,
        count + 100,
        str(count),
        ha="center"
    )

plt.show()
```

```python
fig, axes = plt.subplots(
    2,
    5,
    figsize=(15, 7)
)

for row, cls in enumerate(classes):

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    files = os.listdir(class_path)[:5]

    for col, file in enumerate(files):

        image_path = os.path.join(
            class_path,
            file
        )

        image = Image.open(image_path)

        axes[row, col].imshow(image)
        axes[row, col].set_title(cls)
        axes[row, col].axis("off")

plt.tight_layout()
plt.show()
```

### Image dimensions

```python
image_dimensions = []

for cls in classes:

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    for file in os.listdir(class_path):

        image_path = os.path.join(
            class_path,
            file
        )

        try:
            with Image.open(image_path) as image:
                image_dimensions.append({
                    "class": cls,
                    "filename": file,
                    "width": image.width,
                    "height": image.height
                })

        except Exception:
            pass

dimensions_df = pd.DataFrame(image_dimensions)

print("Unique dimensions:")
print(
    dimensions_df[
        ["width", "height"]
    ].value_counts()
)

print("\nNumber of unique dimensions:",
      dimensions_df[["width", "height"]].drop_duplicates().shape[0])
```

### Pixel representation the 10,000 CT images

```python
image_modes = []

for cls in classes:

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    for file in os.listdir(class_path):

        image_path = os.path.join(
            class_path,
            file
        )

        try:
            with Image.open(image_path) as image:
                image_modes.append({
                    "class": cls,
                    "mode": image.mode
                })

        except Exception:
            pass

modes_df = pd.DataFrame(image_modes)

print("Image modes:")
print(modes_df["mode"].value_counts())

print("\nModes by class:")
print(
    pd.crosstab(
        modes_df["class"],
        modes_df["mode"]
    )
)
```

### File formats
Let's check whether all files are PNG/JPEG

```python
file_formats = []

for cls in classes:

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    for file in os.listdir(class_path):

        image_path = os.path.join(
            class_path,
            file
        )

        try:
            with Image.open(image_path) as image:
                file_formats.append({
                    "class": cls,
                    "format": image.format,
                    "filename": file
                })

        except Exception:
            pass

formats_df = pd.DataFrame(file_formats)

print("Image formats:")
print(formats_df["format"].value_counts())

print("\nFormats by class:")
print(
    pd.crosstab(
        formats_df["class"],
        formats_df["format"]
    )
)
```

```python
corrupted_images = []

for cls in classes:

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    for file in os.listdir(class_path):

        image_path = os.path.join(
            class_path,
            file
        )

        try:
            with Image.open(image_path) as image:
                image.verify()

        except Exception as e:
            corrupted_images.append({
                "class": cls,
                "filename": file,
                "error": str(e)
            })

print("Corrupted images:", len(corrupted_images))

if corrupted_images:
    corrupted_df = pd.DataFrame(corrupted_images)
    display(corrupted_df.head())
else:
    print("No corrupted images found.")
```

```python
import hashlib

image_hashes = []

for cls in classes:

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    for file in os.listdir(class_path):

        image_path = os.path.join(
            class_path,
            file
        )

        try:
            with open(image_path, "rb") as f:
                file_hash = hashlib.md5(
                    f.read()
                ).hexdigest()

            image_hashes.append({
                "class": cls,
                "filename": file,
                "path": image_path,
                "hash": file_hash
            })

        except Exception:
            pass

hash_df = pd.DataFrame(image_hashes)

duplicate_hashes = (
    hash_df["hash"]
    .value_counts()
)

duplicate_hashes = duplicate_hashes[
    duplicate_hashes > 1
]

print(
    "Number of duplicate groups:",
    len(duplicate_hashes)
)

print(
    "Images involved in duplicates:",
    duplicate_hashes.sum()
)
```

```python
duplicate_groups = (
    hash_df[
        hash_df["hash"].isin(duplicate_hashes.index)
    ]
    .groupby("hash")
)

cross_class_duplicates = []

for file_hash, group in duplicate_groups:

    unique_classes = group["class"].unique()

    if len(unique_classes) > 1:

        cross_class_duplicates.append(
            group
        )

print(
    "Cross-class duplicate groups:",
    len(cross_class_duplicates)
)

if cross_class_duplicates:

    cross_class_df = pd.concat(
        cross_class_duplicates,
        ignore_index=True
    )

    display(
        cross_class_df[
            ["class", "filename", "path", "hash"]
        ]
    )

else:
    print(
        "No duplicates found across different classes."
    )
```

```python
pixel_stats = []

for cls in classes:

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    for file in os.listdir(class_path):

        image_path = os.path.join(
            class_path,
            file
        )

        try:
            with Image.open(image_path) as image:

                image_array = np.array(image).astype(
                    np.float32
                )

                pixel_stats.append({
                    "class": cls,
                    "filename": file,
                    "mean": image_array.mean(),
                    "std": image_array.std(),
                    "min": image_array.min(),
                    "max": image_array.max()
                })

        except Exception:
            pass

pixel_stats_df = pd.DataFrame(pixel_stats)

print(pixel_stats_df.groupby("class")[[
    "mean",
    "std",
    "min",
    "max"
]].mean())
```

### Visualize the distributions

```python
plt.figure(figsize=(9, 5))

for cls in classes:

    class_data = pixel_stats_df[
        pixel_stats_df["class"] == cls
    ]

    plt.hist(
        class_data["mean"],
        bins=50,
        alpha=0.6,
        label=cls
    )

plt.title("Distribution of Mean Pixel Intensity")
plt.xlabel("Mean Pixel Intensity")
plt.ylabel("Number of Images")
plt.legend()

plt.show()
```

```python
for cls in classes:

    class_data = pixel_stats_df[
        pixel_stats_df["class"] == cls
    ]

    q1 = class_data["mean"].quantile(0.25)
    q3 = class_data["mean"].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = class_data[
        (class_data["mean"] < lower_bound) |
        (class_data["mean"] > upper_bound)
    ]

    print(f"\n{cls}")
    print("Q1:", round(q1, 2))
    print("Q3:", round(q3, 2))
    print("Lower bound:", round(lower_bound, 2))
    print("Upper bound:", round(upper_bound, 2))
    print("Brightness outliers:", len(outliers))
```

```python
fig, axes = plt.subplots(
    2,
    5,
    figsize=(15, 6)
)

for row, cls in enumerate(classes):

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    files = np.random.choice(
        os.listdir(class_path),
        size=5,
        replace=False
    )

    for col, file in enumerate(files):

        image_path = os.path.join(
            class_path,
            file
        )

        image = Image.open(image_path)

        axes[row, col].imshow(image)
        axes[row, col].set_title(cls)
        axes[row, col].axis("off")

plt.tight_layout()
plt.show()
```

```python
channel_check = []

for cls in classes:

    class_path = os.path.join(
        DATASET_PATH,
        cls
    )

    for file in os.listdir(class_path):

        image_path = os.path.join(
            class_path,
            file
        )

        try:
            image = np.array(
                Image.open(image_path)
            )

            red = image[:, :, 0]
            green = image[:, :, 1]
            blue = image[:, :, 2]

            channel_check.append({
                "class": cls,
                "R_equals_G": np.array_equal(red, green),
                "G_equals_B": np.array_equal(green, blue),
                "R_equals_B": np.array_equal(red, blue)
            })

        except Exception:
            pass

channel_df = pd.DataFrame(channel_check)

print("R = G:")
print(channel_df["R_equals_G"].value_counts())

print("\nG = B:")
print(channel_df["G_equals_B"].value_counts())

print("\nR = B:")
print(channel_df["R_equals_B"].value_counts())
```
