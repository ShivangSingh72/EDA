```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import sklearn
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
disease_df = pd.read_csv("../data/Diseases_and_Symptoms_dataset.csv")
```

```python
disease_df.head()
```

## Target and Feature

```python
X = disease_df.drop("diseases", axis = 1)
y = disease_df["diseases"]
```

```python
print("Feature Matrix:", X.shape)
print("Target:", y.shape)
print("Unique Diseases:", y.nunique())
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
```

```python
pd.DataFrame({
    "Disease": label_encoder.classes_,
    "Encoded Label": range(len(label_encoder.classes_))
}).head(10)
```

### Train-Test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,y_encoded,test_size=0.2,
    random_state=42,stratify=y_encoded
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)
```

### Train Decission Tree 

```python
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train , y_train)
```

```python
dt_pred = dt_model.predict(X_test)
```

```python
dt_accuracy = accuracy_score(y_test, dt_pred)
print("Decision Tree Accuracy:", dt_accuracy)
```

```python
print(classification_report(
    y_test,
    dt_pred,
    target_names=label_encoder.classes_
))
```

### Train Random Forest 

```python
rf_model = RandomForestClassifier(n_estimators = 100 , random_state = 42)
rf_model.fit(X_train , y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
rf_accuracy = accuracy_score(rf_pred , y_test)
print("RF accuracy:" , rf_accuracy)
```

```python
print(classification_report(y_test , rf_pred , 
    target_names = label_encoder.classes_))
```

### Compare RF and DT

```python
comparison = pd.DataFrame({
    "Model":[
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy":[
        dt_accuracy,
        rf_accuracy
    ]
})

comparison
```

```python
plt.figure(figsize=(6,4))

sns.barplot(data=comparison,x="Model",y="Accuracy")

plt.title("Model Accuracy Comparison")
plt.ylim(0,1)

plt.show()
```
