# Sampling & Central Limit Theorem (CLT)
### What we will understand
- **Population:** the complete group
- **Sample:** a smaller group selected from the population
- **Sample mean:** the average of one sample
- **Sampling distribution:** the distribution of many sample means
- **CLT:** when the sample size is sufficiently large, the distribution of sample means becomes approximately normal.

**Important:** Sample size and number of samples are different. In our main example, each sample has **100 observations**, and we take **1,000 samples**, giving us **1,000 sample means**.

```python
# Cell 1: Create a population
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Imagine these are spending amounts of 10,000 customers
population = np.random.exponential(scale=2000, size=10000)

print("Population size:", len(population))
print("Population mean: ₹", round(population.mean(), 2))

plt.hist(population, bins=50)
plt.title("Population Distribution")
plt.xlabel("Customer Spending (₹)")
plt.ylabel("Frequency")
plt.show()

# Notice: the population is NOT bell-shaped.
```

```python
#Take ONE sample and calculate its mean
sample_size = 100

sample = np.random.choice(population, size=sample_size, replace=False)
sample_mean = sample.mean()

print("Sample size:", sample_size)
print("Sample mean: ₹", round(sample_mean, 2))
print("\nThis is just ONE sample and ONE sample mean.")
```

```python
#Take MANY samples and calculate the mean of each sample
number_of_samples = 1000
sample_means = []

for _ in range(number_of_samples):
    sample = np.random.choice(population, size=sample_size, replace=False)
    sample_means.append(sample.mean())

print("Number of samples:", number_of_samples)
print("Observations in each sample:", sample_size)
print("Number of sample means:", len(sample_means))

plt.hist(sample_means, bins=30)
plt.title("Sampling Distribution of the Sample Mean")
plt.xlabel("Sample Mean (₹)")
plt.ylabel("Frequency")
plt.show()

print("\nThis is the key CLT idea:")
print("Many samples → many sample means → approximately normal distribution")
```

```python
#See what happens when sample size increases
sample_sizes = [5, 30, 100]

for n in sample_sizes:
    means = []
    for _ in range(1000):
        sample = np.random.choice(population, size=n, replace=False)
        means.append(sample.mean())

    print(f"Sample size = {n:3d} | Mean of sample means = ₹{np.mean(means):.2f} | SD of sample means = ₹{np.std(means):.2f}")

print("\nRemember:")
print("Larger sample size → sample means become more stable → smaller spread")
print("CLT → with sufficiently large random samples, the sample-mean distribution tends toward normality.")
```
